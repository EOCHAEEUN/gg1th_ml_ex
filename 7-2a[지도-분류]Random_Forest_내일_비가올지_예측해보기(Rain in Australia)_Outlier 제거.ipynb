### Outlier(이상치) 제거
- Outlier : 다른 값들과 동떨어진, 지나치게 크거나 작은 값
- 문제점 : 평균/분산을 왜곡 → 모델이 소수의 극단값에 끌려다니며 학습됨
- 판별 기준 : IQR(사분위 범위) 방식
  - Q1(25% 지점), Q3(75% 지점), IQR = Q3 - Q1
  - 정상 범위 : Q1 - 1.5*IQR ~ Q3 + 1.5*IQR
  - 이 범위 밖 = 이상치 (아래 그림의 boxplot 양 끝 점들)

### 이상치 확인하는 함수
- IQR 방식으로 정상 범위를 벗어난 행의 index(행 번호)를 모아서 반환
- 반환값을 `df.drop(index)`에 넣어 해당 행들을 삭제

```python
# 앞에서 df를 여러 번 건드렸으므로 원본 데이터를 다시 깨끗하게 로딩
df = pd.read_csv('datas_ml/weatherAUS.csv')
print(df.shape)
```

```python
# 이상치 확인하는 함수
def get_outlier_indices(data, columns):
    outlier_indices = []   # 이상치에 해당하는 행의 목록 리스트
    for column in columns:
        Q1 = data[column].quantile(0.25) # 25%분위수(1Q)
        Q3 = data[column].quantile(0.75) # 75%분위수(3Q)
        IQR = Q3 - Q1                    # 가운데 50%가 퍼져 있는 폭
        min_value = Q1 - 1.5*IQR         # 정상 범위 하한
        max_value = Q3 + 1.5*IQR         # 정상 범위 상한
        # 하한보다 작거나(|) 상한보다 크면 True인 참/거짓 배열(boolean mask)
        filter = ((data[column] < min_value) | (data[column] > max_value))

        outlier_data = data[column][filter]   # mask가 True인 값만 추출
        outlier_index = outlier_data.index    # 그 값들의 행 번호
        outlier_indices.extend(outlier_index) # 리스트에 이어붙이기(append 아님 주의)

    return outlier_indices
```

```python
# 0이 아닌 강수량 값만으로 찾은 이상치 행을 원본 df에서 삭제할 준비
print('dataframe before removing outlier (Rainfall column) :', df.shape)
# 0을 제외한 데이터(filtered_rainfall) 기준으로 Rainfall 이상치 행 번호 찾기
delete_indices = get_outlier_indices(filtered_rainfall, ['Rainfall'])
# 필요하면 아래 주석을 해제해 삭제할 행 번호를 직접 확인할 수 있음
# print(delete_indices)
```

```python
# drop(행번호, axis=0) : 해당 행 삭제
# reset_index(drop=True) : 삭제로 비어버린 index를 0부터 다시 매김(기존 index는 버림)
df = df.drop(delete_indices, axis=0).reset_index(drop=True)
print('dataframe after removing outlier (Rainfall column) :', df.shape)
```

```python
print('dataframe before removing outlier (etc) :', df.shape)
# 나머지 주요 컬럼들의 이상치 행 번호 찾기
# - 앞에서 reset_index를 했으므로 지금의 df 기준 index가 사용됨
delete_indices = get_outlier_indices(df, ['Humidity9am','Evaporation', 'WindGustSpeed', 'Pressure9am', 'Pressure3pm'])
```

```python
df = df.drop(delete_indices, axis=0).reset_index(drop=True)
print('dataframe after removing outlier (etc) :', df.shape)   # 행 수 줄어든 것 확인
```

## 결측치(null값) 확인 및 처리하기
- 결측치 : 값이 비어 있는 칸(NaN) → 사이킷런 모델은 NaN이 있으면 학습 불가
- 처리 방법
  - 숫자형 : 중앙값(median)으로 채움 → 이상치 영향을 덜 받음
  - 범주형 : 최빈값(mode)으로 채움
  - 정답(target)이 비어 있으면 : 채우지 않고 행 삭제(정답을 지어낼 수 없음)

```python
# isnull() : 비어 있으면 True / mean() : True 비율 = 결측 비율
# sort_values() : 비율 낮은 순 정렬 → 어떤 컬럼이 많이 비었는지 확인
df.isnull().mean().sort_values()
```

### categorical / numerical column 확인하기

```python
print(cat_cols)
```

```python
print(num_cols)
```

### 결측치 개수 확인하기

```python
# isnull()은 결측치를 True로 표시하고, mean()은 True의 비율(결측 비율)을 계산
df[cat_cols].isnull().mean().sort_values()   # 범주형 컬럼 결측 비율
```

```python
# 숫자형도 같은 방식으로 결측 비율을 확인. 값이 클수록 비어 있는 데이터가 많음
df[num_cols].isnull().mean().sort_values()   # 숫자형 컬럼 결측 비율
```

```python
# 숫자형 컬럼 결측치를 중앙값으로 채우기
for col in num_cols:
    if df[col].isnull().mean()>0:       # 결측이 하나라도 있는 컬럼만
        col_median=df[col].median()     # 그 컬럼의 중앙값 계산
        df[col] = df[col].fillna(col_median)   # NaN 자리에 중앙값 대입
```

```python
# 잘 채워졌는지 검증 : 아무것도 출력되지 않으면 숫자형 결측 0개
for col in num_cols:
    if df[col].isnull().mean()>0:
        print(col, (df[col].isnull().mean()))
```

```python
# 범주형 컬럼 결측치를 최빈값(mode)으로 채우기
# - mode()는 Series를 반환하므로 [0]으로 첫 번째 값(가장 많이 나온 값)을 꺼냄
df['WindGustDir'] = df['WindGustDir'].fillna(df['WindGustDir'].mode()[0])
df['WindDir9am'] = df['WindDir9am'].fillna(df['WindDir9am'].mode()[0])
df['WindDir3pm'] = df['WindDir3pm'].fillna(df['WindDir3pm'].mode()[0])
df['RainToday'] = df['RainToday'].fillna(df['RainToday'].mode()[0])
```

# Yes/No 값에 대한 변환
- 모델은 문자열을 이해하지 못함 → 숫자로 바꿔야 함
- No → 0, Yes → 1 (0/1 두 값이므로 이진 분류 문제)

```python
# map : 값을 1:1로 바꿔주는 함수 (딕셔너리에 없는 값은 NaN이 되므로 주의)
df['RainToday'] = df['RainToday'].map({'No': 0, 'Yes': 1})
df['RainTomorrow'] = df['RainTomorrow'].map({'No': 0, 'Yes': 1})

print(df['RainToday'].dtype, df['RainTomorrow'].dtype)   # 숫자형으로 바뀌었는지 확인
```

## categorical value에 one-hot encoding 적용하기
- one-hot encoding : 범주 하나를 0/1 컬럼 하나로 펼치는 방식
  - 예) WindGustDir(16방향) → WindGustDir_N, WindGustDir_NE, ... 16개 컬럼
- 이유 : 방향/지역에 그냥 1,2,3 숫자를 매기면 모델이 '크기 순서'로 오해함
- 부작용 : 컬럼 수가 크게 늘어남

```python
# get_dummies : 지정한 범주형 컬럼들을 0/1 더미 컬럼으로 자동 변환
df = pd.get_dummies(df, columns = ['Location', 'WindGustDir', 'WindDir9am', 'WindDir3pm'])
```

```python
df.head()
```

```python
# one-hot encoding 후에는 범주별 0/1 컬럼이 생겨 열 수가 늘어남
df.shape
```

# 트레이닝 데이터 준비
- X(feature) : 예측에 사용할 입력값들 / y(target) : 맞혀야 할 정답
- train : 모델이 공부할 데이터 / test : 처음 보는 데이터로 채점할 데이터

```python
X = df.drop(['RainTomorrow'], axis=1)   # 정답 컬럼을 뺀 나머지 전부가 입력
y = df['RainTomorrow']                  # 정답 컬럼
```

```python
# X에는 정답(RainTomorrow)이 빠져 있는지 확인
X.head()
```

```python
# y의 앞 5개 정답 확인: No는 0, Yes는 1로 변환돼 있어야 함
y[:5]
```

```python
# train_test_split : 데이터를 학습용/테스트용으로 분리
# - test_size=0.2   : 20%를 테스트용으로
# - stratify=y      : 원본의 Yes/No 비율을 train, test에 동일하게 유지(불균형 데이터에 필수)
# - random_state    : 난수 고정 → 실행할 때마다 같은 결과 재현
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=2024)
```

# ML 학습, 테스트, 평가
- 내일 비가 올지 안올지 예측해보기
- 학습기 : Random Forest
- Random Forest : 결정트리(Decision Tree) 여러 개를 만들어 다수결로 예측하는 앙상블 모델
  - 트리 하나만 쓸 때보다 과적합(overfitting)에 강함

## ML 학습

```python
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier()   # 모델 객체 생성(기본 트리 100개)
rf_clf.fit(X_train, y_train)        # fit : 학습용 데이터로 학습(공부)

y_pred = rf_clf.predict(X_test)     # predict : 테스트 데이터로 예측(시험)
score = accuracy_score(y_test, y_pred)   # 실제 정답 vs 예측 비교 → 정확도
# baseline - 랜덤 포레스트(Random Forest) Accuracy : 0.8533352086922887
print('랜덤 포레스트(Random Forest) Accuracy (After removing Outlier):', score)
```

## ML 모델 평가
- accuracy(정확도)만으로는 부족 → 불균형 데이터에서는 "전부 No"라고 찍어도 점수가 높게 나옴
- precision(정밀도) : 비 온다고 예측한 것 중 실제로 비 온 비율
- recall(재현율) : 실제 비 온 날 중 모델이 맞힌 비율
- f1-score : precision과 recall의 조화평균(둘의 균형)

```python
from sklearn.metrics import classification_report
# classification_report() 결과 출력
# - 클래스(0/1)별 precision, recall, f1-score, support(개수)를 한 번에 표로 보여줌
print(classification_report(y_test, y_pred))
```

### 문제 XGBoost로 모델 만들고 평가하기
- XGBoost : 트리를 하나씩 순차적으로 추가하며 앞선 트리의 오차를 보정하는 부스팅 모델
- Random Forest(병렬·다수결)와 비교해서 어느 쪽이 더 나은지 확인

```python
from xgboost import XGBClassifier

# XGBoost도 여러 결정트리를 순서대로 학습하는 앙상블 분류 모델
# random_state를 고정하면 실행할 때마다 결과를 비교하기 쉬움
xgb_clf = XGBClassifier(random_state=2024)

xgb_clf.fit(X_train, y_train)      # 학습
y_pred = xgb_clf.predict(X_test)   # 예측

accuracy = accuracy_score(y_test, y_pred)
print('accuracy :', accuracy)
```

```python
# 레포트
print(classification_report(y_test, y_pred))
```

## LGBM 분류기
- LightGBM : XGBoost와 같은 부스팅 계열이지만 학습 속도가 빠르고 대용량에 유리
- 사용법(fit → predict)은 사이킷런 계열 모델과 동일 → 모델만 바꿔 끼우면 됨

```python
from lightgbm import LGBMClassifier

# LightGBM은 많은 데이터를 빠르게 학습할 수 있는 트리 기반 분류 모델
lgbm_clf = LGBMClassifier(random_state=2024)
lgbm_clf.fit(X_train, y_train)
y_pred = lgbm_clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print('accuracy :', accuracy)
```

```python
# 레포트
print(classification_report(y_test, y_pred))
```
